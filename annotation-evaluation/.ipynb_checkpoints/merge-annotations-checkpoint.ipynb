```python
# Columns to remove
cols_to_drop = [
    "doc_id",
    "meta/previous_utterance",
    "meta/following_utterance"
]

# Clean all three DataFrames
for df in [thea, julian, lisa]:
    df.drop(columns=cols_to_drop, inplace=True)
    df.rename(columns={"label/0": "label"}, inplace=True)
```

```python
def rename_label(df, annotator):
    """Rename the label column to include the annotator's name."""
    return df.rename(columns={"label": f"label_{annotator}"})
```

```python
from functools import reduce
import pandas as pd

def merge_annotations(dfs):
    """
    Merge annotation DataFrames on all columns except 'label' and 'id'.
    Creates a separate label column for each annotator.
    """

    # Merge on all columns except label and id
    merge_cols = [
        col for col in next(iter(dfs.values())).columns
        if col not in ["label", "id"]
    ]

    renamed_dfs = []

    for name, df in dfs.items():
        temp = (
            df.drop(columns="id")
              .rename(columns={"label": f"label_{name}"})
        )
        renamed_dfs.append(temp)

    # Merge all DataFrames
    merged = reduce(
        lambda left, right: pd.merge(
            left,
            right,
            on=merge_cols,
            how="inner"
        ),
        renamed_dfs
    )

    return merged
```

```python
df_merged = merge_annotations({
    "lisa": lisa,
    "julian": julian,
    "thea": thea
})
```
